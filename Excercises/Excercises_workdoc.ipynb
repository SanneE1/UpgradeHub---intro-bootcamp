**Libraries import**

````python
import pandas as pd
````

## Carga de datos

````python
direccion_nasa = (
    "https://exoplanetarchive.ipac.caltech.edu/TAP/sync?query="
    "select+pl_name,hostname,disc_year,discoverymethod,disc_facility,"
    "pl_orbper,pl_rade,pl_bmasse,pl_eqt,sy_dist,sy_pnum+from+pscomppars&format=csv"
)  

try:
    tabla_exoplanetas = pd.read_csv(direccion_nasa)
    print('Carga de datos exitosa')
except Exception:
    tabla_exoplanetas = pd.read_csv("data/exoplanetas.csv")
    print("loaded csv from file")
````

````python
tabla_exoplanetas.head(3)
df = tabla_exoplanetas
df.columns
````

## Cambiar los nombres de las columnas
 
Los nombres de las columnas vienen en inglés y abreviados (por ejemplo `pl_rade`). Vamos a ponerles nombres
que entendamos. Esta tabla explica qué significa cada una:
 
| Nombre original | Nombre nuevo | Qué es |
|---|---|---|
| `pl_name` | `nombre` | Nombre del planeta |
| `hostname` | `estrella` | Nombre de la estrella alrededor de la que gira |
| `disc_year` | `anio` | Año en que se descubrió |
| `discoverymethod` | `metodo` | Técnica con la que se descubrió |
| `disc_facility` | `telescopio` | Telescopio o misión espacial que lo encontró |
| `pl_orbper` | `periodo_dias` | Cuánto tarda en dar una vuelta a su estrella (su "año"), en días |
| `pl_rade` | `radio_tierras` | Tamaño del planeta comparado con la Tierra (1 = igual que la Tierra, 2 = el doble) |
| `pl_bmasse` | `masa_tierras` | Peso del planeta comparado con la Tierra (1 = igual que la Tierra) |
| `pl_eqt` | `temperatura_k` | Temperatura estimada, en grados Kelvin (0 °C son 273 K) |
| `sy_dist` | `distancia_pc` | Distancia a la Tierra en pársecs (1 pársec = 3,26 años luz) |
| `sy_pnum` | `num_planetas` | Cuántos planetas tiene ese sistema solar |
 
**Cosa nueva**: `.rename(columns={...})` es la herramienta de pandas para renombrar columnas. Le damos una lista de
parejas `"nombre viejo": "nombre nuevo"` entre llaves `{ }`. Esas llaves son un **diccionario**, que veremos con calma más adelante.


````python
df = df.rename(columns={
    "pl_name": "nombre",
    "hostname": "estrella",
    "disc_year": "anio",
    "discoverymethod": "metodo",
    "disc_facility": "telescopio",
    "pl_orbper": "periodo_dias",
    "pl_rade": "radio_tierras",
    "pl_bmasse": "masa_tierras",
    "pl_eqt": "temperatura_k",
    "sy_dist": "distancia_pc",
    "sy_pnum": "num_planetas",
 
})
df.head()
````

````python
df.shape
df.info()
df.describe().round(2)   # sort of like R summary()
````

## Bucles

````python
lista_de_radios = df["radio_tierras"].dropna().to_list()
RADIO_JUPITER = 11.2
numero_de_gigantes = 0
suma_de_radios = 0

for radio in lista_de_radios:
    if radio > RADIO_JUPITER:
        numero_de_gigantes += 1
        suma_de_radios += radio

print(f"number of giants: {numero_de_gigantes}")
print(f"Sum of Radius: {suma_de_radios}")
````

!! Instead of doing `df = df + 1` you can do `df += 1`

¿cuál es el exoplaneta más cercano a la Tierra y a cuántos años luz está? 

````python
exoplaneta_cercano = df.loc[df["distancia_pc"].idxmin()]

nombre = exoplaneta_cercano["nombre"]
distancia_anios_luz = exoplaneta_cercano["distancia_pc"] * 3.26

print(f"El exoplaneta más cercano es {nombre}.")
print(f"Está a aproximadamente {distancia_anios_luz:.2f} años luz.")
````

¿cuál fue el primer exoplaneta descubierto y en qué año?

````python
first_planet = df.loc[df["anio"].idxmin()]
name = first_planet["nombre"]
year = first_planet["anio"]

print(f"El primer exoplaneta descubierto es {name}.")
print(f"Fue descubierto en el año {year}.")
````

escribe una función `planetas_descubiertos_en(anio)` que devuelva cuántos planetas se descubrieron ese año.

````python
def planetas_descubiertos_en(anio):
    df_temp = df[df["anio"] == anio]
    n = df_temp.shape[0]
    return int(n)

print(f"Planetas descubiertos en 2010: {planetas_descubiertos_en(2010)}")
print(f"Planetas descubiertos en 1995: {planetas_descubiertos_en(1995)}")
````
